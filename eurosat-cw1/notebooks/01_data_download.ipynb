## Step 2: Mount Google Drive & Create Shared CW1 Directory

## Step 3: Download EuroSAT via torchvision

```python
from torchvision.datasets import EuroSAT

print('Downloading EuroSAT RGB dataset...')
ds = EuroSAT(root='/content/data', download=True)

print(f'Total images : {len(ds)}')
print(f'Classes ({len(ds.classes)}): {ds.classes}')
# Expected: 27000 images, 10 classes
```

## Step 4: Verify folder structure

```python
import os

root = '/content/data/eurosat/2750'
classes = sorted(os.listdir(root))

print(f'Found {len(classes)} class folders:')
for c in classes:
    n = len(os.listdir(f'{root}/{c}'))
    print(f'  {c}: {n} images')
```

## Step 5: Copy dataset to Google Drive (persistent storage)

```python
print('Copying EuroSAT to Google Drive...')
!cp -r /content/data/eurosat /content/drive/MyDrive/ann_cw1/
print('Done! Dataset saved to Drive.')
```

## Step 6: Record download metadata (T1a)

```python
import json
from datetime import datetime

metadata = {
    'dataset': 'EuroSAT RGB',
    'citation': 'Helber, P., Bischke, B., Dengel, A., Borth, D. (2019). EuroSAT: A Novel Dataset and Deep Learning Benchmark for Land Use and Land Cover Classification. IEEE JSTARS.',
    'source_imagery': 'Sentinel-2 (Copernicus programme)',
    'total_images': 27000,
    'num_classes': 10,
    'image_size': '64x64 RGB',
    'download_method': 'torchvision.datasets.EuroSAT',
    'download_date': datetime.now().isoformat(),
    'classes': ds.classes,
    'ethical_notes': [
        'No identifiable people in satellite patches.',
        'Coverage limited to European regions — may not generalise globally.',
        'Risk of misuse: land-use classification can support unwanted surveillance.',
        'Automated labels could feed planning decisions without human review.'
    ]
}

with open('/content/drive/MyDrive/ann_cw1/dataset_metadata.json', 'w') as f:
    json.dump(metadata, f, indent=2)

print('Metadata saved.')
print(json.dumps(metadata, indent=2))
```
